```python
#There is no mean for customerID cause it is just identifier
df=df.drop('customerID',axis=1)
```

### Missing Values

The 11 missing values belong to new customers with `tenure = 0`, so they can be safely removed.

```python
df=df.dropna(subset=["TotalCharges"])
```

```python
df.duplicated().sum()
```

```python

df=df.drop_duplicates()
df.duplicated().sum()
```

### Duplicate Records

After removing `customerID`, some records became duplicates because they have identical feature values. These duplicates do not provide new information for our ML model, so we can safely remove them.

```python
df[["tenure", "MonthlyCharges", "TotalCharges"]].describe()
```

```python
pd.crosstab(df["Contract"],df["Churn"],normalize="index")*100
```

```python
pd.crosstab(df["Contract"], df["Churn"], normalize="index").plot(
    kind="bar",
    stacked=True
)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract")
plt.ylabel("Proportion")
plt.legend(title="Churn")
plt.show()
```

```python
df.groupby("Churn")["tenure"].mean()
```

```python
df.groupby("Churn")["MonthlyCharges"].mean()
```

### Key Findings

- Month-to-month customers have a much higher churn rate (42.6%) compared to one-year (11.3%) and two-year contracts (2.8%).
- Churned customers have a lower average tenure (18.1 vs. 37.7 months).
- Churned customers have higher average monthly charges (74.60 vs. 61.39).

```python
categorical_columns=df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts())
```

### Conclusion

The dataset was cleaned and explored successfully.

Key findings:

* 26.5% of customers churned, while 73.5% remained.
* Month-to-month customers have the highest churn rate.
* Churned customers have lower average tenure and higher average monthly charges.
* `tenure` and `TotalCharges` have a strong positive correlation (0.83).
* The target variable is imbalanced, so accuracy alone will not be enough to evaluate the models.

The cleaned dataset is ready for preprocessing and machine learning.
